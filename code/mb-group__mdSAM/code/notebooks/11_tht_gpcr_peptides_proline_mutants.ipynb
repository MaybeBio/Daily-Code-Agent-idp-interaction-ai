```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
x = pd.read_csv('GPCR_ThT/Vikas_15_10_22.tsv',delimiter='\t',header=None)
vikas_v2  = pd.read_csv('GPCR_ThT/Vikas_27_10_22.tsv',delimiter='\t',header=None)
```

```python
d =[]
for i in x.index:
    for j in range(1,7):
        if x.loc[i,0].split('_')[-1]=='SHUFFLE':
            d.append({'name':x.loc[i,0].split('_')[0],'AAseq':x.loc[i,0].split('_')[1],'ThT':x.loc[i,j],'ele':'SHUFFLE','date':1})
        if x.loc[i,0].split('_')[-1]=='MUT':
            d.append({'name':x.loc[i,0].split('_')[0],'AAseq':x.loc[i,0].split('_')[1],'ThT':x.loc[i,j],'ele':'MUT','date':1})
        else:
            d.append({'name':x.loc[i,0].split('_')[0],'AAseq':x.loc[i,0].split('_')[1],'ThT':x.loc[i,j],'ele':'WT','date':1})
```

```python
df = pd.DataFrame(d)
```

```python
d =[]
for i in vikas_v2.index:
    for j in range(1,9):
        if vikas_v2.loc[i,0].split('_')[-1]=='SHUFFLE':
            d.append({'name':vikas_v2.loc[i,0].split('_')[0],'AAseq':vikas_v2.loc[i,0].split('_')[1],'ThT':vikas_v2.loc[i,j],'ele':'SHUFFLE','date':2})
        if vikas_v2.loc[i,0].split('_')[-1]=='MUT':
            d.append({'name':vikas_v2.loc[i,0].split('_')[0],'AAseq':vikas_v2.loc[i,0].split('_')[1],'ThT':vikas_v2.loc[i,j],'ele':'MUT','date':2})
        else:
            d.append({'name':vikas_v2.loc[i,0].split('_')[0],'AAseq':vikas_v2.loc[i,0].split('_')[1],'ThT':vikas_v2.loc[i,j],'ele':'WT','date':2})
```
