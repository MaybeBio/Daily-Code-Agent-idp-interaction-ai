```python
d = waltzdb_extend.loc[:,['Classification','WT_Windows_above_thershold']]
```

```python
d.loc[d[d.Classification=='non-amyloid'].index,'class']=0
d.loc[d[d.Classification!='non-amyloid'].index,'class']=1
```

```python
d = d.drop(columns=['Classification'])
```

```python
l = []
for i in range(1,5):
    TN = len((d[(d.WT_Windows_above_thershold.between(0,i-1))&(d['class']==0)]))
    FN = len(d[(d.WT_Windows_above_thershold.between(0,i-1))&(d['class']==1)])
    TP = len(d[(d.WT_Windows_above_thershold.between(i,10))&(d['class']==1)])
    FP = len(d[(d.WT_Windows_above_thershold.between(i,10))&(d['class']==0)])
    print(TN,FN,TP,FP)
    TPR = TP/(TP+FN)
    FPR = FP/(FP+TN)
    print(TPR,FPR)
    l.append({'c':i,'tpr':TPR,'fpr':FPR})
```

```python
ll = pd.DataFrame(l)
sns.set(font_scale=1,style='ticks',rc={'grid.linestyle': '-','grid.linewidth': .5})
fig, ax = plt.subplots(figsize=(2.5, 2.5)) 
sns.scatterplot(x='fpr',y='tpr',data=ll,hue='c',palette='RdGy',s=75,edgecolor=None)
sns.lineplot(x='fpr',y='tpr',data=ll,zorder=0,color='k')
ax.plot(ll.loc[1].fpr,ll.loc[1].tpr, 'bo', markerfacecolor='none', ms=15, markeredgecolor='red',linewidth=20)
# plt.xlabel('$False\;positive\;rate$')
# plt.ylabel('$True\;positive\;rate$')
plt.title('$ROC\;curve$',fontsize=10)
ax.grid()
# plt.plot([0,1],[0,1])
# plt.xlim([0,.5])
# plt.ylim([0.,.5])
plt.xlabel('$False\;positive\;rate$',fontsize=10)
plt.ylabel('$True\;positive\;rate$',fontsize=10)
plt.locator_params(nbins=6)#plt.xscale('log')
plt.legend(title='$Minimum\;No.\;windows$\n$classified\;as \;amyliod$',loc='lower right',title_fontsize=7,fontsize=7,ncol=1)
plt.savefig('ROC_extneded_waltzdb.svg',bbox_inches='tight',transparent=True)
```
