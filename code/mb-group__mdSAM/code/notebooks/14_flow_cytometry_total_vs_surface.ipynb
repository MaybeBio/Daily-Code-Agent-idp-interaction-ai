```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import ttest_ind

names = ['APP','A4','KCNE4','X3CL1','EVI2A','SYCY1','SIGL6','MANS4','CD34','TIMD4']

hue_col = 'ele'
y_col = 'val'
x_col = 'SNA' if 'SNA' in df.columns else 'DNA'

hue_order = ['WT', 'Mut']
palette = ['#B24438', '#949598']

plot_df = df[
    (df['dye'] == 'total') &
    (df['metric'] == 50) &
    (df['rep'].isin([2, 3, 4])) &
    (df['name'].isin(names))
].copy()

def p_to_star(p):
    if pd.isna(p):
        return ''
    elif p < 0.0001:
        return '****'
    elif p < 0.001:
        return '***'
    elif p < 0.01:
        return '**'
    elif p < 0.05:
        return '*'
    else:
        return 'n.s.'

def get_pvalue(a, b):
    a = pd.Series(a).dropna()
    b = pd.Series(b).dropna()

    if len(a) < 2 or len(b) < 2:
        return np.nan

    _, p = ttest_ind(a, b, equal_var=False)
    return p

def add_wt_mut_stats(ax, data, x_order):
    ymin, ymax = ax.get_ylim()
    yrange = ymax - ymin

    if yrange == 0:
        yrange = 1

    bracket_h = 0.035 * yrange
    gap = 0.04 * yrange

    max_anno_y = ymax

    for i, x in enumerate(x_order):
        wt = data[
            (data[x_col] == x) &
            (data[hue_col] == 'WT')
        ][y_col]

        mut = data[
            (data[x_col] == x) &
            (data[hue_col] == 'Mut')
        ][y_col]

        p = get_pvalue(wt, mut)
        label = p_to_star(p)

        if label == '':
            continue

        group_vals = data[data[x_col] == x][y_col].dropna()
        if group_vals.empty:
            continue

        y = group_vals.max() + gap

        x1 = i - 0.20
        x2 = i + 0.20

        ax.plot(
            [x1, x1, x2, x2],
            [y, y + bracket_h, y + bracket_h, y],
            lw=1,
            c='black',
            clip_on=False
        )

        ax.text(
            (x1 + x2) / 2,
            y + bracket_h,
            label,
            ha='center',
            va='bottom',
            fontsize=8,
            clip_on=False
        )

        max_anno_y = max(max_anno_y, y + bracket_h)

    ax.set_ylim(ymin, max_anno_y + 0.18 * yrange)

ncols = 5
nrows = math.ceil(len(names) / ncols)

fig, axes = plt.subplots(
    nrows=nrows,
    ncols=ncols,
    figsize=(ncols * 3.2, nrows * 2.8),
    sharey=False
)

axes = axes.flatten()

legend_handles = None
legend_labels = None

for ax, name in zip(axes, names):
    sub = plot_df[plot_df['name'] == name].copy()

    x_order = list(pd.unique(sub[x_col]))
    x_order = sorted(x_order)

    sns.barplot(
        x=x_col,
        y=y_col,
        data=sub,
        hue=hue_col,
        hue_order=hue_order,
        order=x_order,
        errorbar='se',
        palette=palette,
        ax=ax
    )

    if legend_handles is None:
        legend_handles, legend_labels = ax.get_legend_handles_labels()

    leg = ax.get_legend()
    if leg is not None:
        leg.remove()

    add_wt_mut_stats(ax, sub, x_order)

    ax.set_title(name)
    ax.tick_params(axis='x', rotation=90)
    ax.locator_params(axis='y', nbins=4)
    ax.set_xlabel('')
    ax.set_ylabel(y_col)

for ax in axes[len(names):]:
    ax.axis('off')

fig.legend(
    legend_handles,
    legend_labels,
    title=hue_col,
    loc='upper center',
    ncol=2,
    frameon=False
)

plt.tight_layout(rect=[0, 0, 1, 0.92])
plt.savefig('all_total_subfigures_WT_vs_Mut_stats.svg', bbox_inches='tight', transparent=False)
plt.show()
```

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import ttest_ind

# Change this to your dataframe name
plot_df = df_total.copy()

# Columns in your dataframe
x_col = 'DNA'
y_col = 'Median'
rep_col = 'Rep'
name_col = 'name_short'
hue_col = 'ele'

hue_order = ['WT', 'Mut']
palette = ['#B24438', '#949598']

# Optional: keep only certain reps
# plot_df = plot_df[plot_df[rep_col].isin([2, 3, 4])].copy()

# Optional: remove EMPTY control
plot_df = plot_df[plot_df[name_col] != 'EMPTY'].copy()

# Optional: use a fixed order
names = ['APP','A4','KCNE4','X3CL1','EVI2A','SYCY1','SIGL6','MANS4','CD34','TIMD4']
#names = [n for n in names if n in plot_df[name_col].unique()]
plot_df = plot_df[plot_df.name_short.isin(names)].copy()
# Or plot all name_short values automatically
names = sorted(plot_df[name_col].dropna().unique())

plot_df[y_col] = pd.to_numeric(plot_df[y_col], errors='coerce')
plot_df[x_col] = pd.to_numeric(plot_df[x_col], errors='ignore')

def p_to_star(p):
    if pd.isna(p):
        return ''
    elif p < 0.0001:
        return '****'
    elif p < 0.001:
        return '***'
    elif p < 0.01:
        return '**'
    elif p < 0.05:
        return '*'
    else:
        return 'n.s.'

def get_pvalue(a, b):
    a = pd.Series(a).dropna()
    b = pd.Series(b).dropna()

    if len(a) < 2 or len(b) < 2:
        return np.nan

    _, p = ttest_ind(a, b, equal_var=False)
    return p

def add_wt_mut_stats(ax, data, x_order):
    ymin, ymax = ax.get_ylim()
    yrange = ymax - ymin
    if yrange == 0:
        yrange = 1

    bracket_h = 0.035 * yrange
    gap = 0.06 * yrange
    max_anno_y = ymax

    summary = (
        data
        .groupby([x_col, hue_col])[y_col]
        .agg(['mean', 'sem'])
        .reset_index()
    )

    summary['sem'] = summary['sem'].fillna(0)
    summary['top'] = summary['mean'] + summary['sem']

    for i, dna in enumerate(x_order):
        wt = data[
            (data[x_col] == dna) &
            (data[hue_col] == 'WT')
        ][y_col]

        mut = data[
            (data[x_col] == dna) &
            (data[hue_col] == 'Mut')
        ][y_col]

        p = get_pvalue(wt, mut)
        label = p_to_star(p)

        if label == '':
            continue

        dna_summary = summary[summary[x_col] == dna]

        if dna_summary.empty:
            group_top = data[data[x_col] == dna][y_col].max()
        else:
            group_top = dna_summary['top'].max()

        y = group_top + gap

        x1 = i - 0.20
        x2 = i + 0.20

        ax.plot(
            [x1, x1, x2, x2],
            [y, y + bracket_h, y + bracket_h, y],
            lw=1,
            c='black',
            clip_on=True
        )

        ax.text(
            (x1 + x2) / 2,
            y + bracket_h,
            label,
            ha='center',
            va='bottom',
            fontsize=8,
            clip_on=True
        )

        max_anno_y = max(max_anno_y, y + bracket_h)

    ax.set_ylim(ymin, max_anno_y + 0.20 * yrange)

ncols = 5
nrows = math.ceil(len(names) / ncols)

fig, axes = plt.subplots(
    nrows=nrows,
    ncols=ncols,
    figsize=(ncols * 3.2, nrows * 2.8),
    sharey=False
)

axes = np.array(axes).flatten()

legend_handles = None
legend_labels = None

stats_rows = []

for ax, name in zip(axes, names):
    sub = plot_df[plot_df[name_col] == name].copy()

    x_order = sorted(sub[x_col].dropna().unique())

    sns.barplot(
        x=x_col,
        y=y_col,
        data=sub,
        hue=hue_col,
        hue_order=hue_order,
        order=x_order,
        errorbar='se',
        palette=palette,alpha=1,
        ax=ax
    )

    if legend_handles is None:
        legend_handles, legend_labels = ax.get_legend_handles_labels()

    leg = ax.get_legend()
    if leg is not None:
        leg.remove()

    add_wt_mut_stats(ax, sub, x_order)

    for dna in x_order:
        wt = sub[(sub[x_col] == dna) & (sub[hue_col] == 'WT')][y_col]
        mut = sub[(sub[x_col] == dna) & (sub[hue_col] == 'Mut')][y_col]
        p = get_pvalue(wt, mut)

        stats_rows.append({
            name_col: name,
            x_col: dna,
            'comparison': 'WT vs Mut',
            'pvalue': p,
            'significance': p_to_star(p)
        })

    ax.set_title(name)
    ax.tick_params(axis='x', rotation=90)
    ax.locator_params(axis='y', nbins=4)
    ax.set_xlabel(x_col)
    ax.set_ylabel('MFI')

for ax in axes[len(names):]:
    ax.axis('off')

fig.legend(
    legend_handles,
    legend_labels,
    title=hue_col,
    loc='upper center',
    ncol=2,
    frameon=False
)

plt.tight_layout(rect=[0, 0, 1, 0.92])
plt.savefig('median_DNA_WT_vs_Mut_stats.svg', bbox_inches='tight', transparent=False)
plt.show()

stats_df = pd.DataFrame(stats_rows)
stats_df
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ttest_rel

# ---------------------------------------------------------
# Read data
# ---------------------------------------------------------
file = "20270728_mdSAM_Constructs.xlsx"
sheet = "Surface"          # or "Total"
metric = "Geometric Mean"

df = pd.read_excel(file, sheet_name=sheet)

# Detect WT/MUT column automatically
if {"WT", "MUT"}.issubset(df["Construct"].unique()):
    condition_col = "Construct"
    protein_col = "Element"
else:
    condition_col = "Element"
    protein_col = "Construct"

# Keep only WT/MUT
df = df[df[condition_col].isin(["WT", "MUT"])].copy()

# ---------------------------------------------------------
# Average technical replicates
# ---------------------------------------------------------
avg = (
    df.groupby(
        [protein_col, condition_col, "DNA", "Rep"],
        as_index=False
    )[metric]
    .mean()
)

palette = {
    "WT": "#d62728",
    "MUT": "#7f7f7f"
}

sns.set_theme(style="whitegrid", font_scale=1.2)

results = []

# ---------------------------------------------------------
# Plot each protein
# ---------------------------------------------------------
for protein in sorted(avg[protein_col].unique()):

    sub = avg[avg[protein_col] == protein]

    plt.figure(figsize=(4.5,4))

    ax = sns.pointplot(
        data=sub,
        x="DNA",
        y=metric,
        hue=condition_col,
        estimator="mean",
        errorbar="sd",
        dodge=0.2,
        palette=palette,
        markers=["o","s"],
        linestyles="-",
    )

    sns.stripplot(
        data=sub,
        x="DNA",
        y=metric,
        hue=condition_col,
        dodge=True,
        palette=palette,
        size=7,
        edgecolor="black",
        linewidth=0.8,
        legend=False,
        ax=ax,
    )

    # ----------------------------
    # Paired t-tests
    # ----------------------------
    ymax = sub[metric].max()

    for i, dna in enumerate(sorted(sub["DNA"].unique())):

        tmp = sub[sub["DNA"] == dna]

        paired = (
            tmp.pivot(
                index="Rep",
                columns=condition_col,
                values=metric
            ).dropna()
        )

        if len(paired) >= 2:

            t, p = ttest_rel(
                paired["WT"],
                paired["MUT"]
            )

            results.append({
                "Protein": protein,
                "DNA": dna,
                "N": len(paired),
                "p-value": p
            })

            if p < 0.001:
                txt = "***"
            elif p < 0.01:
                txt = "**"
            elif p < 0.05:
                txt = "*"
            else:
                txt = f"p={p:.2f}"

            ax.text(
                i,
                ymax*1.08,
                txt,
                ha="center",
                fontsize=11
            )

    # Remove duplicate legends
    handles, labels = ax.get_legend_handles_labels()
    ax.legend(handles[:2], labels[:2], title="")

    ax.set_title(protein, fontsize=14, weight="bold")
    ax.set_xlabel("DNA (µg)")
    ax.set_ylabel(metric)

    plt.tight_layout()
    plt.show()

# ---------------------------------------------------------
# Statistics table
# ---------------------------------------------------------
results = pd.DataFrame(results)
print(results)
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import ttest_1samp

# =====================================================
# Settings
# =====================================================
file = "20270728_mdSAM_Constructs.xlsx"
metric = "Geometric Mean"

protein_col = "Construct"
condition_col = "Element"

# =====================================================
# Read data
# =====================================================
total = pd.read_excel(file, sheet_name="Total")
surface = pd.read_excel(file, sheet_name="Surface")

# Keep WT/MUT only
total = total[total[condition_col].isin(["WT", "MUT"])].copy()
surface = surface[surface[condition_col].isin(["WT", "MUT"])].copy()

# =====================================================
# Average technical replicates
# =====================================================
group_cols = [protein_col, condition_col, "DNA", "Rep"]

total = (
    total.groupby(group_cols, as_index=False)[metric]
    .mean()
    .rename(columns={metric: "Total"})
)

surface = (
    surface.groupby(group_cols, as_index=False)[metric]
    .mean()
    .rename(columns={metric: "Surface"})
)

# =====================================================
# Merge
# =====================================================
df = total.merge(surface, on=group_cols)

# Surface / Total
df["Ratio"] = df["Surface"] / df["Total"]

# =====================================================
# Normalize each replicate to WT
# =====================================================
rows = []

for protein in sorted(df[protein_col].unique()):

    for dna in sorted(df["DNA"].unique()):

        tmp = df[
            (df[protein_col] == protein) &
            (df["DNA"] == dna)
        ]

        paired = (
            tmp.pivot_table(
                index="Rep",
                columns=condition_col,
                values="Ratio",
                aggfunc="mean"
            )
        )

        if not {"WT", "MUT"}.issubset(paired.columns):
            continue

        paired = paired[["WT", "MUT"]].dropna().reset_index()

        paired["FoldChange"] = paired["MUT"] / paired["WT"]
        paired["log2FC"] = np.log2(paired["FoldChange"])
        paired["DNA"] = dna
        paired[protein_col] = protein

        rows.append(paired)

fc = pd.concat(rows, ignore_index=True)

# =====================================================
# Plot
# =====================================================
sns.set_theme(style="whitegrid", font_scale=1.2)

summary = []

for protein in sorted(fc[protein_col].unique()):

    sub = fc[fc[protein_col] == protein]

    plt.figure(figsize=(5,4))
    ax = plt.gca()

    # connect biological replicates
    sns.lineplot(
        data=sub,
        x="DNA",
        y="FoldChange",
        hue="Rep",
        estimator=None,
        legend=False,
        lw=1,
        alpha=0.35,
        color="gray",
        ax=ax,
    )

    # individual replicates
    sns.stripplot(
        data=sub,
        x="DNA",
        y="FoldChange",
        color="#1f77b4",
        size=8,
        edgecolor="black",
        linewidth=0.8,
        jitter=0.05,
        ax=ax,
    )

    # Mean ± SEM
    dna_levels = sorted(sub["DNA"].unique())

    for i, dna in enumerate(dna_levels):

        vals = sub.loc[sub["DNA"] == dna, "FoldChange"]

        mean = vals.mean()
        sem = vals.sem()

        ax.errorbar(
            i,
            mean,
            yerr=sem,
            fmt="_",
            color="black",
            lw=3,
            capsize=5,
            markersize=20,
            zorder=5,
        )

        # One-sample t-test on log2FC
        logvals = np.log2(vals)

        if len(logvals) >= 2:
            _, p = ttest_1samp(logvals, 0)
        else:
            p = np.nan

        summary.append({
            "Protein": protein,
            "DNA": dna,
            "N": len(vals),
            "Mean Fold Change": mean,
            "SEM": sem,
            "Mean log2FC": logvals.mean(),
            "p-value": p,
        })

        # significance label
        if np.isfinite(p):
            if p < 0.001:
                txt = "***"
            elif p < 0.01:
                txt = "**"
            elif p < 0.05:
                txt = "*"
            else:
                txt = f"p={p:.2f}"

            ax.text(
                i,
                mean + sem + 0.05,
                txt,
                ha="center",
                fontsize=10,
            )

    ax.axhline(
        1,
        color="red",
        linestyle="--",
        alpha=0.5,
    )

    ax.set_title(protein, fontsize=14, fontweight="bold")
    ax.set_xlabel("DNA (µg)")
    ax.set_ylabel("MUT / WT\n(Surface / Total)")

    sns.despine()

    plt.tight_layout()
    plt.show()

# =====================================================
# Summary table
# =====================================================
summary = pd.DataFrame(summary)

print(summary.round(3))
```

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

# =====================================================
# Settings
# =====================================================
file = "20270728_mdSAM_Constructs.xlsx"
metric = "Geometric Mean"

protein_col = "Construct"
condition_col = "Element"

# =====================================================
# Read Surface data
# =====================================================
df = pd.read_excel(file, sheet_name="Surface")

# WT/MUT only
df = df[df[condition_col].isin(["WT", "MUT"])].copy()

# =====================================================
# Average technical replicates
# =====================================================
df = (
    df.groupby(
        [protein_col, condition_col, "DNA", "Rep"],
        as_index=False
    )[metric]
    .mean()
)

# =====================================================
# Transform variables
# =====================================================
df["logSurface"] = np.log2(df[metric])
df["logDNA"] = np.log10(df["DNA"])

df[condition_col] = pd.Categorical(
    df[condition_col],
    categories=["WT", "MUT"]
)

results = []

# =====================================================
# Mixed model for every protein
# =====================================================
for protein in sorted(df[protein_col].unique()):

    sub = df[df[protein_col] == protein].copy()

    # Skip proteins with too little data
    if sub["Rep"].nunique() < 2:
        continue

    try:

        model = smf.mixedlm(
            f"logSurface ~ {condition_col} * logDNA",
            data=sub,
            groups=sub["Rep"]
        )

        fit = model.fit(
            method="lbfgs",
            reml=True
        )

        print("\n")
        print("=" * 70)
        print(protein)
        print("=" * 70)
        print(fit.summary())

        # Save p-values
        for term, p in fit.pvalues.items():

            results.append({
                "Protein": protein,
                "Term": term,
                "Estimate": fit.params[term],
                "P-value": p
            })

    except Exception as e:

        print(f"{protein}: model failed ({e})")

# =====================================================
# Summary table
# =====================================================
results = pd.DataFrame(results)

print("\n\nSummary\n")
print(results)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ============================================================
# SETTINGS
# ============================================================

file = "20270728_mdSAM_Constructs.xlsx"
metric = "Geometric Mean"

palette = {
    "WT": "#d62728",
    "MUT": "gray"
}

sns.set_theme(
    style="ticks",
    font_scale=1.1,
)
# Mixed model WT coefficients (Element[T.MUT])
mut_stats = {
    "A4": (1.710, 0.001),
    "APP": (1.959, 0.0001),
    "CD34": (-0.041, 0.939),
    "CX3CL1": (0.129, 0.618),
    "ERVW-1": (-0.126, 0.774),
    "EVI2A": (0.009, 0.991),
    "HTR7": (-0.197, 0.338),
    "HTR7_SNAPtm": (1.040, 0.0001),
    "KCNE4": (0.238, 0.250),
    "KCNE4_SNAPtm": (0.962, 0.002),
    "MANS4": (0.514, 0.159),
    "SIGL6": (-0.157, 0.641),
    "TIMD4": (-0.165, 0.758),
}

# ============================================================
# LOAD DATA
# ============================================================

surface = pd.read_excel(file, sheet_name="Surface")
total = pd.read_excel(file, sheet_name="Total")

surface = surface[surface["Element"].isin(["WT", "MUT"])].copy()
total = total[total["Element"].isin(["WT", "MUT"])].copy()

surface = (
    surface
    .groupby(["Construct", "Element", "DNA", "Rep"], as_index=False)[metric]
    .mean()
    .rename(columns={metric: "Surface"})
)

total = (
    total
    .groupby(["Construct", "Element", "DNA", "Rep"], as_index=False)[metric]
    .mean()
    .rename(columns={metric: "Total"})
)

df = surface.merge(
    total,
    on=["Construct", "Element", "DNA", "Rep"]
)

sns.set_theme(style="ticks", font_scale=1.1)

# ============================================================
# FIGURE FUNCTION
# ============================================================

def make_figure(proteins, title):

    fig, axes = plt.subplots(
        nrows=len(proteins),
        ncols=2,
        figsize=(10, 3.7 * len(proteins)),
        constrained_layout=True
    )

    if len(proteins) == 1:
        axes = [axes]

    for i, protein in enumerate(proteins):

        sub = df[df["Construct"] == protein]

        beta, p = mut_stats[protein]

        # --------------------------
        # SURFACE
        # --------------------------

        ax = axes[i][0]

        sns.stripplot(
            data=sub,
            x="DNA",
            y="Surface",
            hue="Element",
            dodge=True,
            palette=palette,
            size=7,
            edgecolor="black",
            linewidth=0.6,
            ax=ax,
        )

        sns.pointplot(
            data=sub,
            x="DNA",
            y="Surface",
            hue="Element",
            dodge=0.4,
            estimator="mean",
            errorbar="se",
            palette=palette,
            markers="_",
            linestyles="-",
            scale=2,
            errwidth=2,
            capsize=.15,
            ax=ax,
        )

        ax.set_yscale("log")

        ax.set_ylabel("Surface")

        ax.set_title(protein, fontsize=13, fontweight="bold")

        ax.text(
            0.98,
            0.97,
            f"β = {beta:.2f}\np = {p:.3g}",
            transform=ax.transAxes,
            ha="right",
            va="top",
            fontsize=10,
            bbox=dict(
                facecolor="white",
                edgecolor="black",
                alpha=0.8
            )
        )

        handles, labels = ax.get_legend_handles_labels()
        ax.legend(handles[:2], labels[:2], title="")

        # --------------------------
        # TOTAL
        # --------------------------

        ax = axes[i][1]

        sns.stripplot(
            data=sub,
            x="DNA",
            y="Total",
            hue="Element",
            dodge=True,
            palette=palette,
            size=7,
            edgecolor="black",
            linewidth=0.6,
            ax=ax,
        )

        sns.pointplot(
            data=sub,
            x="DNA",
            y="Total",
            hue="Element",
            dodge=0.4,
            estimator="mean",
            errorbar="se",
            palette=palette,
            markers="_",
            linestyles="-",
            scale=2,
            errwidth=2,
            capsize=.15,
            ax=ax,
        )

        ax.set_yscale("log")

        ax.set_ylabel("Total")

        handles, labels = ax.get_legend_handles_labels()
        ax.legend(handles[:2], labels[:2], title="")

    fig.suptitle(title, fontsize=18, fontweight="bold")

    #sns.despine()

    plt.show()

# ============================================================
# MAIN FIGURE
# ============================================================

main = [
    "APP",
    "A4",
    "HTR7_SNAPtm",
    "KCNE4_SNAPtm",
]

make_figure(
    main,
    "Proteins with significant WT/MUT effects"
)

# ============================================================
# SUPPLEMENTARY FIGURE
# ============================================================

supp = sorted(
    [x for x in df["Construct"].unique() if x not in main]
)

make_figure(
    supp,
    "Remaining proteins"
)
```
