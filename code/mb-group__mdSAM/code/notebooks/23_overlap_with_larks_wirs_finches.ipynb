```python
#WIRS and mdSAM
a = len(set.intersection(mdSAM_PM,WAVE_PM))
# No WIRS yes mdSAM
b = len(set.intersection(set.difference(PM,WAVE_PM),mdSAM_PM))
#No WIRS No mdSAM
d = len(set.intersection(set.difference(PM,WAVE_PM),set.difference(PM,mdSAM_PM)))
#No mdSAM yes WIRS
c = len(set.intersection(set.difference(PM,mdSAM_PM),WAVE_PM))

table= [[a,b],[c,d]]
print(table)
print(expected_freq(table))
enrich = (table-expected_freq(table))[0][0]
    
print(enrich, stats.fisher_exact(table,alternative='two-sided')[-1])
```

```python
from itertools import chain, combinations
from upsetplot import from_memberships, plot
import matplotlib.pyplot as plt

# Your 4 sets

# Dictionary of all sets
sets = {
    'WIRS': WAVE_MP,
    'mdSAMs': dSAM_MP,
    'LARKs': LARKs_MP,
    'Fuzzy': alex_set
}

# Generate all non-empty combinations of the set names
def all_combinations(iterable):
    return chain.from_iterable(combinations(iterable, r) for r in range(1, len(iterable) + 1))

memberships = []
counts = []

for combo in all_combinations(sets.keys()):
    included = set.intersection(*(sets[k] for k in combo))
    excluded_keys = set(sets.keys()) - set(combo)
    excluded = set.union(*(sets[k] for k in excluded_keys)) if excluded_keys else set()
    
    # Only count elements unique to this intersection (not in other sets)
    unique = included - excluded
    if unique:
        memberships.append(list(combo))
        counts.append(len(unique))

# Create and plot
example = from_memberships(memberships, data=counts)
plot(example, show_counts=True)
plt.savefig('upsetplot_setlogic.svg', bbox_inches='tight', transparent=True)
```
