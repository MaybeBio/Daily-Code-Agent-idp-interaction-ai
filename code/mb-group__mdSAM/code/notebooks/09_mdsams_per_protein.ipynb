```python
import pandas as pd 
from scipy.spatial import distance
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import time
from multiprocessing import Pool
import tqdm
from itertools import combinations
from scipy import stats
import seaborn as sns; sns.set_theme(style='white')
import matplotlib.pyplot as plt
import matplotlib
from matplotlib.colors import LogNorm, Normalize
from matplotlib.ticker import MaxNLocator
from Levenshtein import distance as lev
```

```python
topology_df = pd.read_pickle('../files/hTMs_uniprot_v1.pkl')
dSAMs = pd.read_pickle('../files/dSAM_MPs_v_paper_v2_new_locs.pkl')
```

```python
one = 0
for i in dSAMs.acc.value_counts():
    if i==1:
        one = one+1
two = 0
for i in dSAMs.acc.value_counts():
    if i==2:
        two = two+1   
three = 0
for i in dSAMs.acc.value_counts():
    if i>=3:
        three = three+1 
fig, ax = plt.subplots(figsize=(2.,2.*1.618)) 
pd.DataFrame([{'x':1,'y':one},{'x':2,'y':two},{'x':3,'y':three}])
sns.barplot(x='x',y='y',data=pd.DataFrame([{'x':1,'y':one},{'x':2,'y':two},{'x':3,'y':three}]),lw=0,palette=['#C03830','#C03830','#C03830'])
plt.ylabel('$No.\;Proteins$')
plt.xlabel('$No.mdSAMs\;in\;protein$')
ax.set_xticklabels(['$1$','$2$','$\\geq 3$'])
ax.grid()
plt.savefig('number_of_mdSAMs_per_protein.svg',bbox_inches='tight',transparent=True)
```
