```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
df = pd.read_excel('CD/20260115_Amytracker520.xlsx',header=9)
set(df.Peptide)
```

```python
#DRD1.APP.HTR7.CCR4
```

```python
fig, ax = plt.subplots(figsize=(3,3))
sns.barplot(x='Peptide',y=' Raw Data  (470-15/513-20)',hue_order=['WT','MUT'],palette=['k','r'],fill=False,
            data=df[(df['Concentration (uM)']==100)&(df.Peptide.isin(['APP','DRD1','5HTR7','CCR4','IFNAR2','IFNAR1']))],hue='Element')
plt.axhline(df[(df.Peptide=='Control')&(df['Concentration (uM)']==100)][' Raw Data  (470-15/513-20)'].median(),c='k',ls='--')
plt.title('Amytracker(100uM)')
plt.locator_params(axis='y',nbins=4)
plt.savefig('Amytracker_100uM.svg',bbox_inches='tight',transparent=False)#ThT FI (A.U.)
```

```python
stats.ttest_ind(df[(df.Peptide=='IFNAR2')&(df['Concentration (uM)']==100)&(df.Element=='WT')][' Raw Data  (470-15/513-20)'],
df[(df.Peptide=='IFNAR2')&(df['Concentration (uM)']==100)&(df.Element=='MUT')][' Raw Data  (470-15/513-20)'],alternative='greater')
```
