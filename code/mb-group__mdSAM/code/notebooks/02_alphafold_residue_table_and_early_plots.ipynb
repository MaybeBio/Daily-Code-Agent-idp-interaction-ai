```python
import numpy as np
from matplotlib.patches import PathPatch

def adjust_box_widths(g, fac):
    """
    Adjust the widths of a seaborn-generated boxplot.
    """

    # iterating through Axes instances
    for ax in g.axes:

        # iterating through axes artists:
        for c in ax.get_children():

            # searching for PathPatches
            if isinstance(c, PathPatch):
                # getting current width of box:
                p = c.get_path()
                verts = p.vertices
                verts_sub = verts[:-1]
                xmin = np.min(verts_sub[:, 0])
                xmax = np.max(verts_sub[:, 0])
                xmid = 0.5*(xmin+xmax)
                xhalf = 0.5*(xmax - xmin)

                # setting new width of box
                xmin_new = xmid-fac*xhalf
                xmax_new = xmid+fac*xhalf
                verts_sub[verts_sub[:, 0] == xmin, 0] = xmin_new
                verts_sub[verts_sub[:, 0] == xmax, 0] = xmax_new

                # setting new width of median line
                for l in ax.lines:
                    if np.all(l.get_xdata() == [xmin, xmax]):
                        l.set_xdata([xmin_new, xmax_new])
```

```python
def ranges(nums):
    nums = sorted(set(nums))
    gaps = [[s, e] for s, e in zip(nums, nums[1:]) if s+1 < e]
    edges = iter(nums[:1] + sum(gaps, []) + nums[-1:])
    return list(zip(edges, edges))
def findMiddle(input_list):
    middle = float(len(input_list))/2
    if middle % 2 != 0:
        return input_list[int(middle - .5)]
    else:
        return input_list[int(middle)]
```

```python
def Random_stretch(ensg,i,number_of_seqs,length):
    agg = [i for i in list(range(int(IDRs.loc[i, 'start'] - length), int(IDRs.loc[i, 'end'] + length))) if i > 0]
    # creating a bubule around the aggrgation region
    pro_no_agg = list((set.difference(set(list(range(0, len(AF2[AF2.ENSG == ensg].AAseq.values[0])))),
                                      set(agg))))  # creating a bubule around the random non aggregation region
    pro_no_agg_IDR = set([ j for j in pro_no_agg if int(AF2[AF2.ENSG==ensg].pLDDT.values[0][j])<70])
    l=[]
    #print(ranges(pro_no_agg_IDR))
    for num in ranges(pro_no_agg):
        pool = [ i for i in list(range(num[0],num[1])) if i<num[1]-length]
        for j in pool:
            if len(set.intersection(set(list(range(int(j+(len(agg)/2)-30),int(j+(len(agg)/2)+30)))),pro_no_agg_IDR))>=6:
                rand = list(range(j, j + len(agg)))
                if len(set.intersection(set(rand),
                                    set(agg)))==0:
                    l.append(rand)
    if len(l)>number_of_seqs:
        return(pd.Series(l).sample(number_of_seqs).to_list())
    else:
        return(pd.Series(l).to_list())
        
```

```python
def Get_PTM_pattern(PTM_type='Ubiquitination'):
    #PTM_type = 'Ubiquitination'#Phosphorylation,Acetylation,Ubiquitination,Methylation
    import random
    import os
    half_win = 100
    PTM_enrich = pd.DataFrame()
    PTM_enrich2 = pd.DataFrame()

    c = 0
    PTM_all = pd.DataFrame()
    for i in IDRs[IDRs.ENSG.isin(ENSG_agg)].index:
        ensg = IDRs.loc[i, 'ENSG']
        orginal_index = i
        if os.path.isfile(
                '/Users/aelazar/PycharmProjects/PhaseSep/classifier/notebooks_for_paper/ENSG_PTMs/%s.pkl' % IDRs.loc[
                    i, 'ENSG']) == True:
            agg = [i for i in list(range(int(IDRs.loc[i, 'start'] - half_win), int(IDRs.loc[i, 'end'] + half_win))) if i > 0]
            # creating a bubule around the aggrgation region
            c = c + 1
            if c % 100 == 0:
                print(c)
            PTM = pd.read_pickle(
                '/Users/aelazar/PycharmProjects/PhaseSep/classifier/notebooks_for_paper/ENSG_PTMs/%s.pkl' % IDRs.loc[
                    i, 'ENSG'])  # from  pyiptmnet api; https://research.bioinformatics.udel.edu/iptmnet/
            PTM_sites = [int(j[1:]) for j in
                         PTM[(PTM.ptm_type == '%s' % PTM_type) & (PTM.score >= 1)].site.dropna().values]
            PTM_enrich2 = PTM_enrich2.append(
                {'box': 'box', 'val': len(set.intersection(set(PTM_sites), set(agg))), 'type': 'Aggregation',
                 'region': agg}, ignore_index=True)
            for slide in enumerate(agg):
                mid = agg.index(findMiddle(agg))
                sliding_win = list(range(slide[1]-3, slide[1] + 3))
                agg_per_sliding = len(set.intersection(set(sliding_win), set(PTM_sites)))
                PTM_enrich = PTM_enrich.append(
                    {'box': 'box', 'ENSG': ensg, 'win': slide[0] - mid, 'val': agg_per_sliding, 'type': 'Aggregation',
                     'AAseq': IDRs.loc[i, 'AAseq'], 'orginal_index': orginal_index},
                    ignore_index=True)  #:len(set.intersection(set(agg),set(PTM_sites)))

            rand_samp = Random_stretch(ensg,i,20,half_win)
            if not rand_samp:
                print('EMPTY')
                continue
            else:
                for ran in rand_samp:
                    PTM_enrich2 = PTM_enrich2.append(
                        {'box': 'box', 'val': len(set.intersection(set(PTM_sites), set(ran))), 'type': 'Random',
                         'region': rand}, ignore_index=True)
                    for slide in enumerate(ran):
                        mid = ran.index(findMiddle(ran))
                        sliding_win = list(range(slide[1]-3, slide[1] + 3))
                        agg_per_sliding = len(set.intersection(set(sliding_win), set(PTM_sites)))
                        PTM_enrich = PTM_enrich.append({'box': 'box', 'ENSG': ensg, 'win': slide[0] - mid, 'val': agg_per_sliding, 'type': 'Random',
                             'AAseq': IDRs.loc[i, 'AAseq'], 'orginal_index': orginal_index}, ignore_index=True)
            PTM_all = pd.concat([PTM_all, PTM])
        else:
            print('NO PTM data')
    PTM_enrich.to_pickle('PTM_enrich_100_%s.pkl' % PTM_type)
    PTM_enrich2.to_pickle('PTM_enrich2_100_%s.pkl' % PTM_type)
    return PTM_enrich2,PTM_enrich

```

```python
def plotplattern(df,name):
    f, axs = plt.subplots(nrows=1, ncols=1, sharey=False,figsize=(13, 8))
    sns.lineplot(x='win',y='val',data=df,hue='type',palette=['r','k'],estimator='mean',ci=95)
    #sns.lineplot(x='win',y='val',data=PTM_enrich[(PTM_enrich.type!='Agg')],color='k',estimator='mean')
    #sns.lineplot(x='win',y='val',data=PTM_enrich[(PTM_enrich.type=='Agg')],color='r',estimator='mean')
    print(stats.ttest_ind(df[df.type=='Aggregation'].val,df[df.type!='Aggregation'].val))
    #plt.ylim([0,.4])
    plt.legend(title='$Type\;of\; strech$',loc='best',frameon=False,fontsize=14)
    plt.xlim([-60,60])
    plt.ylim([-.0,.1])
    plt.title("$%s,\;n=%d$"%(name,len(set(df.orginal_index.dropna()))),fontsize=30)
    #plt.text(1, 0.05,"$n=%d$"%(len(set(df.orginal_index))),fontsize=30)
    plt.ylabel('$No.\;%s\;sites$'%name)
    plt.xlabel('$Position\;relative\;to\;aggregation\;strech(N2C)$')
    plt.savefig('%s_enrich.png'%name)
```
