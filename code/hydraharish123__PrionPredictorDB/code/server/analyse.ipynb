```python
from scipy.stats import mannwhitneyu

yes_degree = network_df.loc[
    network_df["Aggregating-like"] == "Yes",
    "node_degree"
]

no_degree = network_df.loc[
    network_df["Aggregating-like"] == "No",
    "node_degree"
]

stat, p = mannwhitneyu(
    yes_degree,
    no_degree,
    alternative="two-sided"
)

print("Mann–Whitney U:", stat)
print("p-value:", p)
```

```python
n1 = len(yes_degree)
n2 = len(no_degree)

U = stat

rank_biserial = 2 * U / (n1 * n2) - 1

print("Rank-biserial correlation:", rank_biserial)
```

```python
The aggregation-positive proteins have significantly different network degrees from the aggregation-negative proteins, and the positive rank-biserial effect indicates that the aggregation-positive group tends to have higher degree.

A p-value of ~`3 × 10⁻⁵` means the observed separation would be quite unlikely under the null hypothesis that the two groups have the same degree distribution.

The effect size of 0.366 is not enormous, but it is substantial enough to be biologically interesting.
```

```python
from scipy.stats import fisher_exact
from statsmodels.stats.multitest import multipletests

results = []

total_yes = (network_df["Aggregating-like"] == "Yes").sum()
total_no = (network_df["Aggregating-like"] == "No").sum()

for community_id, group in network_df.dropna(
    subset=["community"]
).groupby("community"):

    yes = (group["Aggregating-like"] == "Yes").sum()
    no = (group["Aggregating-like"] == "No").sum()

    outside_yes = total_yes - yes
    outside_no = total_no - no

    table = [
        [yes, no],
        [outside_yes, outside_no]
    ]

    odds_ratio, p_value = fisher_exact(
        table,
        alternative="greater"
    )

    results.append({
        "community": community_id,
        "n_proteins": len(group),
        "n_aggregating": yes,
        "aggregation_fraction": yes / len(group),
        "odds_ratio": odds_ratio,
        "p_value": p_value
    })

community_enrichment = pd.DataFrame(results)

community_enrichment["FDR"] = multipletests(
    community_enrichment["p_value"],
    method="fdr_bh"
)[1]

community_enrichment = community_enrichment.sort_values("FDR")
```

```python
import numpy as np
import pandas as pd

perm_df = network_df.copy()

# Create degree bins based on degree quantiles
perm_df["degree_bin"] = pd.qcut(
    perm_df["node_degree"],
    q=5,
    duplicates="drop"
)

print(
    perm_df.groupby("degree_bin", observed=True)
    .agg(
        n=("gene", "size"),
        yes=("Aggregating-like", lambda x: (x == "Yes").sum())
    )
)
```

```python
community_12_genes = set(
    network_df.loc[
        network_df["community"] == 12,
        "gene"
    ]
)

observed_yes = network_df[
    network_df["gene"].isin(community_12_genes)
]["Aggregating-like"].eq("Yes").sum()

observed_n = len(community_12_genes)

observed_fraction = observed_yes / observed_n

print("Community 12 size:", observed_n)
print("Observed aggregation-positive:", observed_yes)
print("Observed fraction:", observed_fraction)
```

```python
rng = np.random.default_rng(42)

n_permutations = 10000

permuted_counts = np.zeros(n_permutations)

for i in range(n_permutations):

    shuffled_labels = perm_df["Aggregating-like"].copy()

    for _, idx in perm_df.groupby(
        "degree_bin",
        observed=True
    ).groups.items():

        shuffled_labels.loc[idx] = rng.permutation(
            shuffled_labels.loc[idx].values
        )

    community_mask = perm_df["gene"].isin(
        community_12_genes
    )

    permuted_counts[i] = (
        shuffled_labels[community_mask] == "Yes"
    ).sum()
```

```python
empirical_p = (
    np.sum(permuted_counts >= observed_yes) + 1
) / (
    n_permutations + 1
)

print("Observed:", observed_yes)
print("Mean expected:", permuted_counts.mean())
print("95th percentile:", np.percentile(permuted_counts, 95))
print("99th percentile:", np.percentile(permuted_counts, 99))
print("Empirical p-value:", empirical_p)
```

```python
import numpy as np
import pandas as pd
from scipy.stats import mannwhitneyu

yes = network_df.loc[
    network_df["Aggregating-like"] == "Yes", "node_degree"
]

no = network_df.loc[
    network_df["Aggregating-like"] == "No", "node_degree"
]

print("Aggregation = Yes")
print("n:", len(yes))
print("median:", yes.median())
print("mean:", yes.mean())
print("IQR:", yes.quantile([0.25, 0.75]).tolist())

print("\nAggregation = No")
print("n:", len(no))
print("median:", no.median())
print("mean:", no.mean())
print("IQR:", no.quantile([0.25, 0.75]).tolist())

print("\nThreshold enrichment")

for threshold in [5, 10, 15, 20]:
    yes_frac = (yes >= threshold).mean()
    no_frac = (no >= threshold).mean()

    print(
        f"Degree >= {threshold}: "
        f"Yes = {yes_frac:.3f}, "
        f"No = {no_frac:.3f}"
    )
```

```python
import statsmodels.api as sm

model_df = network_df.copy()

model_df["aggregation_binary"] = (
    model_df["Aggregating-like"] == "Yes"
).astype(int)

model_df["log_degree"] = np.log1p(model_df["node_degree"])

X = sm.add_constant(model_df["log_degree"])
y = model_df["aggregation_binary"]

model = sm.Logit(y, X).fit()

print(model.summary())

odds_ratio = np.exp(model.params["log_degree"])
print("\nOdds ratio per unit increase in log(1 + degree):",
      odds_ratio)
```

```python
Aggregation-like proteins exhibited significantly greater network connectivity than non-aggregating proteins, with aggregation status positively associated with log-transformed degree (logistic regression, OR = 2.09, p < 10⁻⁵).
```
